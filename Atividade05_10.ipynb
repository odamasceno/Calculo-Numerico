### **Parte 3 — Implementação numérica em Python**

```python
import numpy as np

# Parte 3: Exercício 4 - Implementação da Eliminação de Gauss
def eliminacao_gauss(A, b):
    # Formar a matriz aumentada
    n = len(b)
    Ab = np.concatenate((A, b.reshape(n, 1)), axis=1).astype(float)

    # Eliminar os elementos abaixo dos pivôs
    for i in range(n):
        for j in range(i + 1, n):
            fator = Ab[j, i] / Ab[i, i]
            Ab[j] = Ab[j] - fator * Ab[i]

    # Substituição
    x = np.zeros(n)
    for i in range(n - 1, -1, -1):
        x[i] = (Ab[i, -1] - np.dot(Ab[i, i+1:n], x[i+1:n])) / Ab[i, i]

    return x

# Teste com o sistema do exercício 2
A_ex2 = np.array([[2, 1, -1], [-3, -1, 2], [-2, 1, 2]])
b_ex2 = np.array([8, -11, -3])

solucao_manual = eliminacao_gauss(A_ex2, b_ex2)
solucao_numpy = np.linalg.solve(A_ex2, b_ex2)

print("Parte 3: Exercício 4:")
print("Solução via algoritmo próprio:", solucao_manual)
print("Solução via numpy.linalg.solve:", solucao_numpy)
```

A solução própria coincide perfeitamente com a do NumPy e com o cálculo manual (x=2, y=3, z=-1).
